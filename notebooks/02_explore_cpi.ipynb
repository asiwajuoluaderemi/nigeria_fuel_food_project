# CPI / Food Index Extraction

Extracts the national monthly food index from NBS CPI "Tables" files.

```python
import pandas as pd
```

## Extraction Function

Unlike petrol/diesel, a single CPI file's Table1 sheet contains many months as rows already. This function pulls the "Food" monthly index column, using the August 2026 file since it covers the longest span (November 2023 – August 2026) in one sheet.

```python
def load_august_cpi_file(file_path):
    df = pd.read_excel(file_path, sheet_name="Table1", header=[2,3])
    food_monthly = df[("Food ", "Monthly")]

    food_df = pd.DataFrame({
        "month_label": df[df.columns[1]],
        "food_index": df[("Food ", "Monthly")]
    })
    
    food_df = food_df.iloc[11:].reset_index(drop=True)
    food_df["month_label"] = ["2023-11", "2023-12", "2024-01", "2024-02", "2024-03",
                                "2024-04", "2024-05", "2024-06",
                                "2024-07", "2024-08", "2024-09",
                                "2024-10", "2024-11", "2024-12",
                                "2025-01", "2025-02", "2025-03",
                                "2025-04", "2025-05", "2025-06",
                                "2025-07", "2025-08", "2025-09",
                                "2025-10", "2025-11", "2025-12",
                                "2026-01", "2026-02", "2026-03",
                                "2026-04", "2026-05", "2026-06",
                                "2026-07", "2026-08"]

    return food_df 
```

## Save Master Table

Saving the extracted food index series to cpi_food_master.csv.

```python
august_df = load_august_cpi_file("../data/raw/cpi/cpi_Report_August_2026.xlsx")
august_df
```
